# Task 7 – Exploratory PCA

This notebook performs a small exploratory Principal Component Analysis (PCA) using Python.

Variables used: `VehPower`, `VehAge`, `DrivAge`, `BonusMalus`, and `Density`.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

df = pd.read_csv('freMTPL2freq.csv')
df.head()
```

## Select numeric predictor variables

`ClaimNb` is excluded because it is the response variable. `Exposure` is excluded because it represents time at risk rather than a general risk characteristic.

```python
pca_data = df[['VehPower', 'VehAge', 'DrivAge', 'BonusMalus', 'Density']].copy()
pca_data.describe()
```

## Correlation matrix

This checks whether the numeric variables are sufficiently related for PCA to provide useful dimensionality reduction.

```python
correlation_matrix = pca_data.corr()
correlation_matrix
```

## Standardize the variables

PCA is performed on standardized variables because the variables are measured on different scales.

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(pca_data)
```

## Run PCA

```python
pca = PCA()
pca_result = pca.fit_transform(scaled_data)

explained_variance = pca.explained_variance_ratio_
cumulative_variance = explained_variance.cumsum()

pd.DataFrame({
    'Principal Component': [f'PC{i+1}' for i in range(len(explained_variance))],
    'Explained Variance': explained_variance,
    'Cumulative Variance': cumulative_variance
})
```

## Component loadings

The loadings show how strongly each original variable contributes to each principal component.

```python
loadings = pd.DataFrame(
    pca.components_.T,
    index=pca_data.columns,
    columns=[f'PC{i+1}' for i in range(len(pca_data.columns))]
)

loadings
```

## Scree plot

The scree plot shows how much variance is explained by each principal component.

```python
plt.figure(figsize=(7, 4))
plt.plot(
    range(1, len(explained_variance) + 1),
    explained_variance,
    marker='o'
)
plt.xlabel('Principal Component')
plt.ylabel('Explained Variance Ratio')
plt.title('Scree Plot of Principal Components')
plt.xticks(range(1, len(explained_variance) + 1))
plt.grid(True, alpha=0.3)
plt.show()
```

## Cumulative variance plot

```python
plt.figure(figsize=(7, 4))
plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    marker='o'
)
plt.axhline(y=0.80, linestyle='--')
plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('Cumulative Explained Variance')
plt.xticks(range(1, len(cumulative_variance) + 1))
plt.grid(True, alpha=0.3)
plt.show()
```
